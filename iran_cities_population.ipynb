# Iran Cities Population

## Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Data

```python
cities = pd.read_csv("iran_top20_cities.csv")
```

## Visualization

```python
from mpl_toolkits.basemap import Basemap
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10, 8))
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])

m = Basemap(llcrnrlon=44., llcrnrlat=25., urcrnrlon=64., urcrnrlat=40.,
            rsphere=(6378137.00, 6356752.3142),
            resolution='l', projection='merc',
            lat_0=32., lon_0=53., lat_ts=32.)

m.drawcoastlines()
m.drawcountries()
m.fillcontinents(color='lightgreen', lake_color='lightblue')
m.drawmapboundary(fill_color='lightblue')

# cities
for i, row in cities.iterrows():
    x, y = m(row["lon"], row["lat"])
    m.scatter(x, y, s=row["population"]/10000, color="red", alpha=0.6, zorder=5)
    ax.annotate(row["city"], (x, y), fontsize=8, zorder=6)

m.drawparallels(np.arange(25, 40, 5), labels=[1,0,0,0])
m.drawmeridians(np.arange(44, 64, 5), labels=[0,0,0,1])

ax.set_title("Iran Cities Population")
plt.show()
```
