# Titanic Data Analysis

## Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Data Extraction

```python
from sklearn.datasets import fetch_openml
titanic = fetch_openml("titanic", version=1, as_frame=True).frame #dataframe
```

```python
titanic.head()
titanic.info()
titanic.describe()
```

## Data Cleaning

### Missing Values

```python

titanic.drop(columns=["cabin", "boat", "body", "home.dest"], inplace=True) #too much null
titanic["fare"] = titanic["fare"].fillna(titanic["fare"].median()) #numeric
titanic["embarked"] = titanic["embarked"].fillna(titanic["embarked"].mode()[0]) #categorical
titanic["age"] = titanic["age"].fillna(titanic["age"].median()) #numeric
```

```python
titanic.isna().sum()
```

### Wrong Format

```python
titanic.dtypes
```

```python
titanic["age"] = titanic["age"].astype(int)
titanic["survived"] = titanic["survived"].astype(bool)
titanic.dtypes
```

```python
titanic.describe()
```
